```python
class python_class:
    def first_method(self):
        print("This is my first method")
    
    def sum_of_numbers(self, a, b):
        print("Sum of 2 numbers is", a + b)

first_object = python_class()
first_object.first_method()

first_object.sum_of_numbers(5, 10)
```

```python
# constructors: special function that are automatically called when an object of a class is created.
# In python, the constructor is defined using the __init__() method.

class python_class:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print("Name:", self.name)
        print("Age:", self.age)
```

```python
class test:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        # print("This is a constructor method")

    def normal_method(self):
        print("my name is " + self.name + " and my age is " + str(self.age))
        # print("my age is " + str(self.age))
        print("This is a normal method")

    def test_function(self):
        print("This is a test function")
        
obj = test("krish", 20)
obj.normal_method()
obj.test_function()
```

```python
# self is a refernce to the current instance of the calss it allows use ot access attributes and 
# methods within the class
class calas:
    def __init__(self, subject):
        self.subject = subject 
    def first_year(self):
        print(f"first year of {self.subject}")

BSCS_Course = calas("Computer Science")
BSCS_Course.first_year()
BBA_Course = calas("Finance")
BBA_Course.first_year()
```
