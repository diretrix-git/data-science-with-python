# Data Science with Python Assignment 1
### Step 1: Read the CSV File

```python
import pandas as pd

df = pd.read_csv('pandas_dataset.csv')
df.head()
```

### Step 2: Rename Columns

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

### Step 3: Drop Unnecessary Rows

```python
initial_shape = df.shape
df = df.dropna(how='all')
df = df.reset_index(drop=True)
```

### Step 4: Drop Columns

```python
df = df.drop(columns=['order_date'])
```

### Step 5: Count Missing Values

```python
missing_values = df.isnull().sum()
print("Missing values:\n", missing_values)
```

### Step 6: Handle Missing Values

```python
df = df.dropna(subset=['sales'])
df['profit'] = df['profit'].fillna(df['profit'].mean())
df['discount'] = df['discount'].fillna(df['discount'].median())
```

### Step 7: Handle Missing Categorical Data

```python
df['customer_name'] = df['customer_name'].fillna('Unknown')
```

### Step 8: Detect and Remove Duplicates

```python
duplicates_count = df.duplicated(subset=['order_id']).sum()
print(f"Duplicates found: {duplicates_count}")
df = df.drop_duplicates(subset=['order_id'])
```

### Step 9: Filtering and creating new columns

```python
high_price = df[df['unit_price'] > 20000]
display(high_price)

filtered_orders = df[(df['unit_price'] > 10000) & (df['status'] == 'Completed')]
display(filtered_orders[['customer_name', 'category', 'status']])

df['total_amount'] = (df['quantity'] * df['unit_price']) - df['discount']
df['customer_type'] = df['quantity'].apply(lambda x: 'Bulk Buyer' if x >= 3 else 'Regular Buyer')
```

### Step 10: Final Clean Dataset

```python
print(f"Shape before cleaning: {initial_shape}")
print(f"Shape after cleaning: {df.shape}")
df.head(10)
```
